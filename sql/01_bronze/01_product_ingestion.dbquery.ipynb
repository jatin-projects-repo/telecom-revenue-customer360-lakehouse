```sql



CREATE OR REPLACE TABLE
telecom_lakehouse.bronze.products_raw
USING DELTA
AS

SELECT
    CONCAT(
        'PROD',
        LPAD(CAST(id AS STRING), 4, '0')
    ) AS product_id,

    CASE
        WHEN PMOD(id, 3) = 0 THEN 'POSTPAID'
        WHEN PMOD(id, 3) = 1 THEN 'PREPAID'
        ELSE 'BROADBAND'
    END AS product_type,

    CONCAT('Plan_', id) AS product_name,

    CASE
        WHEN PMOD(id, 3) = 0 THEN 499
        WHEN PMOD(id, 3) = 1 THEN 299
        ELSE 799
    END AS monthly_price,

    CASE
        WHEN PMOD(id, 4) = 0 THEN 'BASIC'
        WHEN PMOD(id, 4) = 1 THEN 'STANDARD'
        WHEN PMOD(id, 4) = 2 THEN 'PREMIUM'
        ELSE 'ULTIMATE'
    END AS product_category,

    'ACTIVE' AS product_status,

    'PRODUCT_MASTER' AS source_system,

    CURRENT_TIMESTAMP() AS ingestion_timestamp

FROM RANGE(1, 501);
```
