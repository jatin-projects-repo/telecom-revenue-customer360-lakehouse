```sql
CREATE OR REPLACE TABLE bronze.crm_customers_raw
USING DELTA
COMMENT 'Synthetic raw customer data from telecom CRM'
AS

SELECT

    CONCAT(
        'CUST',
        LPAD(CAST(id AS STRING), 7, '0')
    ) AS customer_id,

    CONCAT(
        'Customer_',
        LPAD(CAST(id AS STRING), 7, '0')
    ) AS customer_name,

    CASE
        WHEN PMOD(id, 23) = 0 THEN NULL
        ELSE CONCAT(
            'customer',
            CAST(id AS STRING),
            '@telecomdemo.com'
        )
    END AS email,

    CONCAT(
        '+91',
        CAST(6000000000 + id AS STRING)
    ) AS phone_number,

    CASE
        WHEN PMOD(id, 3) = 0 THEN 'Retail'
        WHEN PMOD(id, 3) = 1 THEN 'SME'
        ELSE 'Enterprise'
    END AS customer_segment,

    CASE
        WHEN PMOD(id, 5) = 0 THEN 'North'
        WHEN PMOD(id, 5) = 1 THEN 'South'
        WHEN PMOD(id, 5) = 2 THEN 'East'
        WHEN PMOD(id, 5) = 3 THEN 'West'
        ELSE 'Central'
    END AS region,

    CASE
        WHEN PMOD(id, 97) = 0 THEN 'UNKNOWN'
        WHEN PMOD(id, 10) < 7 THEN 'ACTIVE'
        WHEN PMOD(id, 10) < 9 THEN 'SUSPENDED'
        ELSE 'TERMINATED'
    END AS customer_status,

    CAST(
        DATE_ADD(
            DATE'2022-01-01',
            CAST(PMOD(id * 17, 1400) AS INT)
        ) AS STRING
    ) AS signup_date,

    'CRM' AS source_system,

    'BATCH_001' AS ingest_batch_id,

    CURRENT_TIMESTAMP() AS ingestion_timestamp

FROM RANGE(1, 10001);
```
